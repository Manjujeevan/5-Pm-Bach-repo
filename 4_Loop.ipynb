### **Python Loop**  
**While loop**  *condition statements*  
*Syntax*  
**while \<condition\>:**

```python
a = 1  # Initilization/Start

while a<=10:  # Condition/End 
    print(a,end=' ')
    a += 1  #Incrementation/Jump
print("\nLoop Over")
```

```python
a = 0
while a<=100:
    print(a,end=' ')
    a += 10
print("\nLoop Over")
```

### **break** – *stop the loop*


```python

# Infinite Loop
a = 1
while True:  # a>0
    print(a,end=' ')
    a+=1
```

```python
a = 1

while a>=0:
    if a == 20:
        
        break
    print(a,end=' ')
    a += 1
```

### **continue** – *go back to the top of my loop*


```python
a = 1

while a<=20:
    if a % 3 == 0:
        print(a,end=' ')
        a += 1

        continue

    
    a += 1
```

```python
""" 
12345
12345
12345
12345
12345
"""

a = 1
while a<=5:
    b = 1
    while b<=5:
        print(b,end='')
        b += 1
    print()
    a += 1
print("Loop Over")
```

### **For loop**  

**range()** – *is a python standard function used for numbers*

*Syntax*  
**for \<variable_Name\> in \<str\\List\\Tuple\\Set\\Dict\>:**

```python
for x in "Hello world":    # Fetches Letter by letter
    print(x,end='')
print("\nLoop Over")
```

```python
mylist = [100,30.55,"hello",True,"world"]  

for x in mylist:   # Fetches Item by Item
    print(x,end=',')
```

### **range()** – *is a python standard function used for numbers*


```python
for x in range(100+1):  # end before
    print(x,end=' ')
```

```python
for x in range(50,100+1): # start with end before
    print(x,end=' ')
```

```python
for x in range(50,100+1,10):  # Jump
    print(x,end=' ')
```

```python
for x in range(100,0,-10):
    print(x,end=' ')
```

### **for with else**  

* If Loop completes successfully, else will worke
* If loop is  unsuccessfully , else will not worke 


```python
for x in "hello world":
    print(x,end='')

else:
    print("\nThe above loop complieted successfully")
```

```python
for x in "hello world":
    if x == ' ':
        break
    print(x,end='')

else:
    print("\nThe above loop complieted successfully")
```

```python
# find if a number is divisible by 5 in the list

mylist = [2,6,9,10,15,11,20,16,97]

for each in mylist:
    if each % 5 == 0:
        print(each,"Divisible By 5")
        break
else:
    print("Number divisible by 5 not found")
```

### **Nested Loop**

```python
""" 
12345
12345
12345
12345
12345
 """


for x in range(5):
    for y in range(1,6):    
        print(y,end='')
    print()
```

```python
row = int(input("Enter a Row  Number :"))
collumn = int(input("Enter a Collumn  Number :"))

for x in range(row):
    for y in range(1,collumn+1):    
        print(y,end='')
    print()
```
