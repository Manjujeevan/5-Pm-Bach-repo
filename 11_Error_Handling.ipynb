**Dividing a number with Error Handling.**

```python
print("Few lines above")

try:
    num1 = int(input("Enter a Dividend number :"))
    print(f"The User Entered Dividend Number is = {num1}")

    num2 = int(input("Enter a Divisor number :"))
    print(f"The User Entered Divisor Number is = {num2}")

    quotient = num1/num2
    print(f"The Quotient of {num1}/{num2} = {quotient}")

except ValueError:
    print("You gave wrong input type")

except ZeroDivisionError:
    print("Dont divide by zero")

print("Few lines below")
```

```python
print("Few lines above")

while True:
    try:
        num1 = int(input("Enter a Dividend number:"))
        print(f"The User Entered Dividend Number is = {num1}")

        num2 = int(input("Enter a Divisor number:"))
        print(f"The User Entered Divisor Number is = {num2}")

        quotient = num1/num2
        print(f"The Quotient of {num1}/{num2} = {quotient}")
        break
        
    except ValueError:
        print("You gave wrong input type, Please enter the numbers again")
        
    except ZeroDivisionError:
        print("Dont divide by zero,  Please enter the numbers again")
        
print("Few lines below")
```

The finally block lets you execute code, regardless of the result of the  try and except blocks.


```python
print("Few lines above")

try:
    num1 = int(input("Enter a number :"))
    print(f"The User Entered Number is = {num1}")
except:
    print("Error handling is done here")
finally:
    print("This line will always work")
    
print("Few lines below")
```

```python
while True:
    try:
        num = int(input("Enter a number inbetween 10-50:"))
        if num<10:
            raise ValueError
        elif num>50:
            raise ZeroDivisionError
        else:
            print(f"Thank you for entering the = {num}")
            break

    except ValueError:
        print("Please Enter a numbaer greater than or equal to 10")
    except ZeroDivisionError:
        print("Please Enter a number lesser than or equal to 50")
```

```python
try:
    num = int(input("Enter a number inbetween 10-50:"))
    if num<10:
        raise ValueError("Please Enter a number greater than or equal to 10")
    elif num>50:
        raise ZeroDivisionError("Please Enter a number lesser than or equal to 50")
    else:
        print(f"Thank you for entering the {num}")
except ValueError as x:
    print(x)
except ZeroDivisionError as y:
    print(y)
```

```python
try:
    print("Outer try block")

    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print(result)

    except ValueError:
        print("Inner Error: Invalid input (not a number)")

    except ZeroDivisionError:
        print("Inner Error: Cannot divide by zero")

except Exception as e:
    print("Outer Error:", e)
```
