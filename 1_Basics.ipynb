### **Escape sequences**  

</center>


* \ - for newline in editor
* \\\ - prints \   
* \’ – prints ‘  
* \” – prints “  
* \n – prints a new line
* \t – prints a tab


```python
print("Hello \\world\\")  # \world\

print("10\\100")
print('I\'m Moon')
print("\"Hello\" \"World\"")  # "Hello""World"

print("Hello\nWorld") # \n => New line
print("hello\tworld")  # \t => tab
```
