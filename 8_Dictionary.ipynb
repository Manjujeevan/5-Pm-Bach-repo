### **Dictionary Functions**  

```python
mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
     
}

print(mydict,type(mydict))
print(mydict["Name"])
print(mydict["Cours"])


mydict["Name"]="Thenmozhi"
mydict["Age"]= 35
print(mydict)
```

```python
# keys()


mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
     
}

print(mydict)
print(list(mydict.keys()))
print()

for each in mydict.keys():
    print(each)
```

```python
# values()

mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
     
}

print(mydict)
print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```

```python
# items()

mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)
print(list(mydict.items()))
print()

for each in mydict.items():
    print(each)
```

```python
mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)
print(list(mydict.items()))
print()

for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

```python
# list = pop()
# list = pop(index)

mylist = [100,30.55,True,"hello",30.55,"hai"]
print(mylist)
mylist.pop(4)
print(mylist)
```

```python
# set = pop()

myset = {100,30.55,True,"hello",30.55,"hai"}
print(myset)
```

```python
# pop(key)

mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)

mydict.pop("City")
print(mydict)
```

```python
# popitem()

mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)
mydict.popitem()
mydict.popitem()
mydict.popitem()
print(mydict)
```

```python
# clear()
mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)

mydict.clear()
print(mydict)
```

```python
# update()
mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)

newdit = {"Name":"Kumar","City":"Mumbai","Cours":"Java","Age": 40}
mydict.update(newdit)
print(mydict)
```

```python
# get()

mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)
print(mydict[500])
print(mydict[600])
print("Hello Python")
```

```python
mydict ={
    "Name": "Honey",
    "City": "Chennai",
   "Cours": "Python",
      500 : 50.55,
}

print(mydict)
print(mydict.get(500))
print(mydict.get(600))
print("Hello Python")
```

```python
# copy()

x = {"A":100,"B":200,"C":300}
y = {"A":100,"B":200,"C":300}

x["D"]=400
y["E"]=500

print(x,type(x),id(x))
print(y,type(y),id(y))
```

```python

x = {"A":100,"B":200,"C":300}
y = x.copy()

x["D"]=400

y["E"]=500


print(x,type(x),id(x))
print(y,type(y),id(y))
```

### **Nested Dictionary**

```python
mydict = {
    "Name": "Manjunathan",
    "City":["Chennai","Kovai","Mumbai"],
    "Year":(2000,2010,2020)
}

print(mydict,type(mydict))
print(mydict.keys())
print('--------------')
print(mydict.get("Name"))
print(mydict.get("City"))
print(mydict.get("City")[0])
print(mydict.get("City")[1])
print('--------------')
print(mydict.get("Year")[2])
print('--------------')

print(mydict.get("City")[2][3:])
```

```python

x = {
    "menu": {
        "id": "file",
        "value": "File",
        "popup": {
            "menuitem": [
                {"value": "New", "onclick": "CreateNewDoc()"},
                {"value": "Open", "onclick": "OpenDoc()"},
                {"value": "Close", "onclick": "CloseDoc()"}
            ]
        }
    }
}


print(x,type(x))
print(x.keys())
print('---------------')
print(x.get("menu")['popup'])
print(x.get("menu")['popup']['menuitem'])
print('---------------')
print(x.get("menu")['popup']['menuitem'][0])
print(x.get("menu")['popup']['menuitem'][1])
print(x.get("menu")['popup']['menuitem'][2])
print('---------------')
print(x.get("menu")['popup']['menuitem'][1]["onclick"])
print(x.get("menu")['popup']['menuitem'][2]["value"])
```
