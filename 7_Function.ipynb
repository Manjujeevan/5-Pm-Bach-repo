### **Arbitrary Arguments**  

**Types of arguments**   
* *args - argument stores in tuple   
* **kwargs - keyworded argument stores in dictionary


```python
def myfunc(*x):   # *args - argument stores in tuple
    print(x,type(x))

myfunc()
myfunc(10)
myfunc("Hai")
myfunc(True)
myfunc(50.55)
print('-----------------------------')
myfunc(10,"hai",True,50.55)
```

```python
def myfunc(**x):   # **kwargs - keyworded argument stores in dictionary

    print(x,type(x))

myfunc(a = 10)
myfunc(name = "Jeevan",age = 15, city = "Kanchi")
```

### **Function overloading**  

```python
def myfunc(x):
    return x

def myfunc(x,y):
    return x+y

def myfunc(x,y,z):
    return x+y+z


print(myfunc(20,60,40))
```
