### **File Handling**

**"r" = Read Mode**

```python
myfile = open("C:/Users/manju/Desktop/Nisha.txt")

print(myfile.tell()) # 0

content = myfile.read(100) 
print(content)
print(myfile.tell())

content = myfile.read(200) 
print(content)
print(myfile.tell())

myfile.seek(0)
print(myfile.tell())
content = myfile.read(200) 
print(content)
print(myfile.tell())

myfile.close()
```

**With - Keyword**

```python
with open("C:\\Users\\manju\\Desktop\\Nisha.txt") as myfile:

    read_line = myfile.readline()
    print(read_line)
   
```

```python
with open("C:\\Users\\manju\\Desktop\\Nisha.txt") as myfile:

    read_lines = myfile.readlines()
    print(read_lines)
    print("--------------------------")
   
    print(read_lines[2])

    print("--------------------------")

    for each in read_lines:
        print(each)
```

**"a" - Append Mode**

```python
with open("C:\\Users\\manju\\Desktop\\Nisha.txt",'a') as myfile:

    My_write = myfile.write("\nWelcom to Gen AI")
```

**"w" - Write Mode**

```python
with open("C:\\Users\\manju\\Desktop\\Nisha.txt",'w') as myfile:
     My_write = myfile.write("Manju Data")
     My_write = myfile.write("\nName : Manjujeevan,")
     My_write = myfile.write("\nCity: Chennai,")
     My_write = myfile.write("\nCours: Python")

    
```
